```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate 
from langchain_core.tools import Tool

# 🔐 Load API key from .env
load_dotenv(dotenv_path=".env")
google_api_key=os.getenv("GOOGLE_API_KEY")

# 🔸 Step 1: Initialize the Gemini LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0,
)

# 🔸 Step 2: Create a Chat Prompt Template
prompt = ChatPromptTemplate.from_template(
    "Answer clearly: {question}"
)

# 🔸 Step 3: Create the LCEL Chain
qa_chain = prompt | llm

# 🔸Step 4: Wrap LCEL Chain as a Tool
qa_tool = Tool(
    name="Simple QA",
    func=lambda question: qa_chain.invoke(
        {"question": question}
    ).content,
    description="A basic LLM chain that answers clear questions"
)

# 🔸 Step 5: Use the Tool
query = "What is LangGraph in LangChain?"

answer = qa_tool.invoke(query)

# 🖨️ Output
print("\nUser Question:", query)
print("\nTool Answer:", answer)
```
