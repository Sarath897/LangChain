```python
import os
from google import genai
from dotenv import load_dotenv

load_dotenv()

# Pass the string explicitly rather than relying on automated injection
client = genai.Client(api_key=os.environ.get("GOOGLE_API_KEY"))

# 1. Initialize a chat session
chat = client.chats.create(model='gemini-3.6-flash')

# 2. Send the message through the chat session
response = chat.send_message('What is LangGraph in LangChain?')

print(response.text)
```

```python
import os
from dotenv import load_dotenv 
from langchain_google_genai import ChatGoogleGenerativeAI 
from langchain_core.prompts import ChatPromptTemplate 

load_dotenv() 

# 🛠️ Changed the model to a generative chat model
llm = ChatGoogleGenerativeAI( 
    model="gemini-3.6-flash", 
    google_api_key=os.getenv("GOOGLE_API_KEY"), 
    temperature=0 
) 

# 🔸 Create Prompt Template 
prompt = ChatPromptTemplate.from_template( "Answer clearly: {question}" ) 

# 🔸 Create Chain (LCEL) 
qa_chain = prompt | llm 

# 🚀 Run a query 
query = "What is LangGraph in LangChain?" 
response = qa_chain.invoke( {"question": query} ) 

# 🖨️ Output 
print("\nUser Question:", query) 
print("\nLLM Answer:", response.content)
```
